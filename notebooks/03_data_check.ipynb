```python
print("Duplicate check pee coint")
print("="*20)
all_clear =True
for symbol in sorted(df['symbol'].unique()):
    coin =df[df['symbol']== symbol]
    dups =coin.duplicated(subset ='date').sum()
    #dups = df[df['symbol']==coin].duplicated(subset=["date"]).sum()
    status ='ok'if dups == 0 else f"{dups} "
    print(f"{symbol}: dups:{status}")
    if dups>0:
        print(f"{coin}: {dups}")
        all_clear=False
    print('='*20)
    print("all coints are clear" if all_clear else "some have duplicated")
```
