```python
import sys
sys.path.insert(0, '..')  # so we can import configs

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from configs import config

# Load the data
df = pd.read_parquet(config.RAW_DATA_FILE)

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print()
print("Coins:", df['symbol'].nunique())
print(df['symbol'].value_counts())
```

```python
btc = df[df['symbol'] == 'BTC'].copy()
btc['date'] = pd.to_datetime(btc['date'])
btc['day'] = btc['date'].dt.date

# Count rows per day
rows_per_day = btc.groupby('day').size()

print("Rows per day (BTC):")
print(rows_per_day.describe())
print()
print("First 5 days:")
print(rows_per_day.head())
```
